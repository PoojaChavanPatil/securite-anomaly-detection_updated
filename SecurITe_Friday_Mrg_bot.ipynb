## flag rows with negatives in real problem columns 

```python
bad_columns_bot = ['Flow IAT Min', 'Flow Duration','Flow Bytes/s', 'Flow Packets/s', 'Flow IAT Mean', 'Flow IAT Max']
row_is_bad_negative = (df_bot[bad_columns_bot] < 0).any(axis=1)
row_is_bad_negative.sum()
```

## flag rows with exactly zero duration 

```python
row_is_bad_zero_duration = (df_bot['Flow Duration'] == 0)
row_is_bad_zero_duration.sum()
```

## flag rows with impossible rate (huge OR inf)

```python
#Building bad_columns_bot and checking rows
row_is_bad_rate_bot = (
    (df_bot['Flow Bytes/s'] > 1e9) 
    | np.isinf(df_bot['Flow Bytes/s']) 
    | np.isinf(df_bot['Flow Packets/s'])
)
row_is_bad_rate_bot.sum()
```

```python
# Combine all three rules with OR — a row is "bad" if ANY rule flags it.

row_is_bad_bot = row_is_bad_negative | row_is_bad_zero_duration | row_is_bad_rate_bot
print("Total bad rows (combined, duplicates removed):", row_is_bad_bot.sum())

# Cross-check: before dropping anything, see what Label these bad rows actually have
df_bot[row_is_bad_bot]['Label'].value_counts()
```

```python
import numpy as np
import pandas as pd
from sklearn.feature_selection import mutual_info_classif
```

```python
# Step 1: Split into the two groups we're comparing
benign_rows = df_bot_clean[df_bot_clean['Label'] == 'BENIGN']
bot_rows = df_bot_clean[df_bot_clean['Label'] == 'Bot']


# Step 2: Get every numeric column (this is what we're testing for separation)
numeric_columns = df_bot_clean.select_dtypes(include=[np.number]).columns.tolist()

```

### Cohen's d

```python
def calculate_cohens_d(column_name):
    mean_benign, mean_bot = benign_rows[column_name].mean(), bot_rows[column_name].mean()
    std_benign, std_bot = benign_rows[column_name].std(), bot_rows[column_name].std()
    n_benign, n_bot = len(benign_rows), len(bot_rows)

    # combined spreadof both the groups together
    pooled_std = np.sqrt(((n_benign - 1) * std_benign**2 + (n_bot - 1) * std_bot**2) / (n_benign + n_bot - 2))

    return 0 if pooled_std == 0 else (mean_bot - mean_benign) / pooled_std

#collecting results
cohens_d_series = pd.Series({col: calculate_cohens_d(col) for col in numeric_columns}, name="Cohen's d")
```

### Mutual Information

```python
label_as_numbers = (df_bot_clean['Label'] == 'Bot').astype(int)   # Bot = 1, BENIGN = 0
X = df_bot_clean[numeric_columns].fillna(0)                        # safety net, just in case

mutual_info_scores = mutual_info_classif(X, label_as_numbers, random_state=0)
mutual_info_series = pd.Series(mutual_info_scores, index=numeric_columns, name="Mutual Info")
```

```python
### Putting both results side by side in one table so we can compare
```

```python
feature_comparison = pd.DataFrame({
    "Cohen's d": cohens_d_series,
    "Mutual Info": mutual_info_series
})
# Sort by the SIZE of Cohen's d
feature_comparison["cohens_d_size"] = feature_comparison["Cohen's d"].abs()
feature_comparison_sorted = feature_comparison.sort_values("cohens_d_size", ascending=False)
feature_comparison_sorted = feature_comparison_sorted.drop(columns="cohens_d_size")

feature_comparison_sorted.head(15)
```

### Plot1 — Destination Port: BENIGN vs Bot

```python
import matplotlib.pyplot as plt


# Step 1: find the top 5 most common ports in each group
top_ports_benign = benign_rows['Destination Port'].value_counts().head(5)
top_ports_bot = bot_rows['Destination Port'].value_counts().head(5)

# combine both lists into one
ports_to_show = sorted(set(top_ports_benign.index) | set(top_ports_bot.index))

# for each port, calculate what PERCENT of each group used it.

benign_percent = benign_rows['Destination Port'].value_counts(normalize=True).reindex(ports_to_show).fillna(0) * 100
bot_percent = bot_rows['Destination Port'].value_counts(normalize=True).reindex(ports_to_show).fillna(0) * 100

# plot BENIGN and Bot side by side for each port
x_positions = range(len(ports_to_show))
bar_width = 0.35
colors = {'BENIGN': '#2a78d6', 'Bot': '#eb6834'}

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar([i - bar_width/2 for i in x_positions], benign_percent, width=bar_width, label='BENIGN', color=colors['BENIGN'])
ax.bar([i + bar_width/2 for i in x_positions], bot_percent, width=bar_width, label='Bot', color=colors['Bot'])

ax.set_xticks(list(x_positions))
ax.set_xticklabels([str(port) for port in ports_to_show])
ax.set_xlabel('Destination Port')
ax.set_ylabel('% of rows in that group')
ax.set_title('Most Common Destination Ports: BENIGN vs Bot')
ax.legend()

plt.tight_layout()
plt.show()
```

Nearly (about 64%) of all Bot rows connect to port 8080.
Begign spreads across multiple legitimate services
