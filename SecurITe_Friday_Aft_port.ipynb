## Buiding the bad rows list here


```python
bad_columns_portscan = ['Flow IAT Min', 'Flow Duration', 'Flow Bytes/s', 
                          'Flow Packets/s', 'Flow IAT Mean', 'Flow IAT Max']

row_is_bad_negative = (df_portscan[bad_columns_portscan] < 0).any(axis=1)
row_is_bad_negative.sum()
```

176 rows with genuinely negative values

## Checking for zero duration rows now 

```python
row_is_bad_zero_duration = (df_portscan['Flow Duration'] == 0)
row_is_bad_zero_duration.sum()
```

371 rows with exactly zero duration

```python
row_is_bad_portscan = row_is_bad_negative | row_is_bad_zero_duration
row_is_bad_portscan.sum()
```

547 combined (176 + 371, no overlap)

```python
(df_portscan['Flow Bytes/s'] > 1e9).sum()
```

```python
## I need to check whether they're mostly BENIGN mostly PortScan before i decide to drop
```

```python
# check which Label these zero-duration rows actually belong to
df_portscan[df_portscan['Flow Duration'] == 0]['Label'].value_counts()
```

```python
# check the same for the impossible-rate rows
df_portscan[df_portscan['Flow Bytes/s'] > 1e9]['Label'].value_counts()
```

```python
# look at the actual PortScan rows with zero duration - what do their other columns look like?
df_portscan[(df_portscan['Flow Duration'] == 0) & (df_portscan['Label'] == 'PortScan')][
    ['Flow Duration', 'Total Fwd Packets', 'Total Backward Packets', 'Flow Bytes/s', 'Flow Packets/s']
].head(15)
```

```python
row_is_bad_rate_portscan = (df_portscan['Flow Bytes/s'] > 1e9) | np.isinf(df_portscan['Flow Bytes/s']) | np.isinf(df_portscan['Flow Packets/s'])

row_is_bad_final_portscan = row_is_bad_negative | row_is_bad_zero_duration | row_is_bad_rate_portscan
row_is_bad_final_portscan.sum()
```

```python
df_portscan_clean = df_portscan[~row_is_bad_final_portscan].copy()
df_portscan_clean.shape
df_portscan_clean['Label'].value_counts()
```

### Plot 1: Flow Duration

```python
import plotly.express as px

fig = px.histogram(df_portscan_clean, x='Flow Duration', color='Label', 
                    nbins=50, log_y=True, barmode='overlay', opacity=0.6,
                    title='Friday - Flow Duration: BENIGN vs PortScan')
fig.update_layout(xaxis_title='Flow Duration (microseconds)', yaxis_title='Number of Flows (log scale)')
fig.show()
```

### Plot 2: Total Fwd Packets

```python
fig = px.histogram(df_portscan_clean, x='Total Fwd Packets', color='Label',
                    nbins=50, log_y=True, barmode='overlay', opacity=0.6,
                    title='Friday - Total Fwd Packets: BENIGN vs PortScan')
fig.update_layout(xaxis_title='Total Forward Packets', yaxis_title='Number of Flows (log scale)')
fig.show()
```

```python
fig = px.box(df_portscan_clean, x='Label', y='Flow Duration', color='Label',
             title='Friday - Flow Duration Spread: BENIGN vs PortScan', log_y=True)
fig.show()
```

### Plot 4: Scatter — Flow Duration vs Total Fwd Packets,

```python
sample_benign = df_portscan_clean[df_portscan_clean['Label'] == 'BENIGN'].sample(5000, random_state=42)
sample_portscan = df_portscan_clean[df_portscan_clean['Label'] == 'PortScan'].sample(5000, random_state=42)
sample_balanced = pd.concat([sample_benign, sample_portscan])

fig = px.scatter(sample_balanced, x='Flow Duration', y='Total Fwd Packets', color='Label',
                  opacity=0.5, log_x=True, log_y=True,
                  color_discrete_map={'BENIGN': 'lightblue', 'PortScan': 'red'},
                  title='Friday - Duration vs Packet Count (log scale, balanced sample)')
fig.update_traces(marker=dict(size=6))
fig.show()
```

PortScan (red/orange): almost entirely locked onto the bottom row, Total Fwd Packets = 1,
duration range of roughly 10 to 1,000 microseconds
spread across the entire chart

```python
flag_summary = sample_balanced.groupby(['Label', 'SYN Flag Count', 'ACK Flag Count']).size().reset_index(name='Count')
flag_summary['Combo'] = 'SYN=' + flag_summary['SYN Flag Count'].astype(str) + ', ACK=' + flag_summary['ACK Flag Count'].astype(str)

fig = px.bar(flag_summary, x='Combo', y='Count', color='Label', barmode='group',
             title='Friday - SYN/ACK Flag Combinations: BENIGN vs PortScan')
fig.show()
```

PortScan essentially never has ACK=1, while a real chunk of BENIGN traffic does
