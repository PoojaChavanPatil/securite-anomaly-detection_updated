## identify numeric column

```python
# same comprehensive negative-value scan like Monday
numeric_cols = df_DDos.select_dtypes(include=[np.number]).columns
neg_counts = (df_DDos[numeric_cols] < 0).sum()
neg_counts[neg_counts > 0].sort_values(ascending=False)
```

## flag rows with negatives in real problem columns 

```python
bad_columns_DDos = ['Flow IAT Min', 'Fwd IAT Min', 'Flow Duration', 
                     'Flow Bytes/s', 'Flow Packets/s', 'Flow IAT Mean', 'Flow IAT Max']
row_is_bad_negative = (df_DDos[bad_columns_DDos] < 0).any(axis=1)
row_is_bad_negative.sum()
```

## flag rows with exactly zero duration 

```python
row_is_bad_zero_duration = (df_DDos['Flow Duration'] == 0)
row_is_bad_zero_duration.sum()
```

## flag rows with impossible rate (huge OR inf)

```python
#Building bad_columns_DDos and checking rows
row_is_bad_rate_DDos = (
    (df_DDos['Flow Bytes/s'] > 1e9) 
    | np.isinf(df_DDos['Flow Bytes/s']) 
    | np.isinf(df_DDos['Flow Packets/s'])
)
row_is_bad_rate_DDos.sum()
```

## Combine ALL THREE into one final flag

```python

row_is_bad_final_DDos = row_is_bad_negative | row_is_bad_zero_duration | row_is_bad_rate_DDos
row_is_bad_final_DDos.sum()
```

```python
## Drop the bad rows, create clean dataframe
```

```python
df_DDos_clean = df_DDos[~row_is_bad_final_DDos].copy()
df_DDos_clean.shape
```

## Mutual Information
To take care of the non-linear relationships

```python
import pandas as pd
from sklearn.feature_selection import mutual_info_classif

# 1. Encode the target column (converts 'BENIGN', 'DDoS', etc. into integers)
df_DDos_clean['Label_num'] = df_DDos_clean['Label'].astype('category').cat.codes

# 2. Compute mutual information
mi_scores = mutual_info_classif(df_DDos_clean[numeric_cols], df_DDos_clean['Label_num'], random_state=42)
mi_ranked = pd.Series(mi_scores, index=numeric_cols).sort_values(ascending=False)
mi_ranked.head(15)
```

```python
## Correlation 
```

```python
corr_raw = df_DDos_clean[numeric_cols].corrwith(df_DDos_clean['Label_num']).sort_values(key=abs, ascending=False)
corr_raw.head(15)
```

Close to +1 → rows tend to be DDoS (since DDoS = 1)
Close to -1 → rows tend to be BENIGN (since BENIGN = 0)
Close to 0 → this column tells me basically nothing about whether a row is DDoS or BENIGN

1.larger backward packet lengths are associated with DDoS traffic.
2.higher port numbers lean toward BENIGN traffic in this dataset.


## Plot_1 balanced sample scatter plot - does DDoS form a distinct cluster?

```python
import plotly.express as px

# take an equal-sized sample from each label so the plot isn't dominated by one class
benign_sample = df_DDos_clean[df_DDos_clean['Label'] == 'BENIGN'].sample(10000, random_state=42)
ddos_sample = df_DDos_clean[df_DDos_clean['Label'] == 'DDoS'].sample(10000, random_state=42)
balanced_sample = pd.concat([benign_sample, ddos_sample])

fig = px.scatter(
    balanced_sample, 
    x='Flow Duration', y='Flow Bytes/s', 
    color='Label', 
    log_x=True, log_y=True,
    title='DDoS vs BENIGN - Duration vs Byte Rate (log scale)'
)
fig.show()
```

Straight diagonal lines of red dots basically means the ddos sending fixed total amount of data just spread over diff time.

DDoS (red) does not form a separate visual cluster from BENIGN (blue) the two overlap heavily across most of the plot.
