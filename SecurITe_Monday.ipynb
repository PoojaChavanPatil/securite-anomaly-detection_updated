The 6 families — this is the mental map

1. Size/Volume family — "how much data moved?"
Total Fwd Packet, Total Bwd Packets, Total Length of Fwd/Bwd Packet, Packet Length Min/Max/Mean/Std, Average Packet Size, Subflow Fwd/Bwd Packets/Bytes
→ In plain words: how many packets, how big were they. A DDoS often means huge packet counts; a PortScan often means tiny packet sizes since there's barely any real data exchanged.

2. Timing family — "how fast or slow did things happen?"
Flow Duration, Flow IAT Mean/Std/Max/Min, Fwd/Bwd IAT ..., Active Min/Mean/Max/Std, Idle Min/Mean/Max/Std
→ In plain words: "IAT" = Inter-Arrival Time = the gap between one packet and the next. This family tells you the rhythm of the conversation — steady and human-like, or rapid-fire and robotic (a classic attack signature). Active/Idle tell you if the connection is bursty (talk, pause, talk, pause) vs continuous.

3. Rate family — "how fast was data flowing, per second?"
Flow Bytes/s, Flow Packets/s, Fwd Packets/s, Bwd Packets/s
→ In plain words: speed of data transfer. Floods (DDoS) push these numbers extremely high.

4. TCP Flag family — "what kind of TCP signals were sent?"
SYN Flag Count, FIN Flag Count, RST Flag Count, PSH Flag Count, ACK Flag Count, URG Flag Count, CWR/ECE Flag Count, Fwd/Bwd PSH/URG Flags
→ In plain words: each flag is a small signal in the TCP handshake/conversation (you already know SYN = "let's start," FIN = "let's end," ACK = "got it," RST = "abort/reset"). Attacks often show abnormal flag patterns — e.g., tons of SYNs with barely any ACKs back (a classic SYN-flood signature).

5. Header/Structure family — "connection setup details"
Fwd/Bwd Header Length, Fwd/Bwd Init Win bytes, Fwd Seg Size Min, Fwd Act Data Pkts
→ In plain words: technical bookkeeping about how the connection was set up. Useful but more subtle — I'd treat this family as "nice to know," not priority right now.

6. Bulk/Ratio family — "bigger-picture shape of the transfer"
Down/Up Ratio, Fwd/Bwd Bytes/Bulk Avg, Fwd/Bwd Packet/Bulk Avg, Fwd/Bwd Bulk Rate Avg, Fwd/Bwd Segment Size Avg
→ In plain words: is this mostly one-directional (attacker pushing, target silent) or balanced (normal back-and-forth)?

## Cleaning 

Monday's file had two related data quality problems, both traced back to the same root cause — a bug in CICFlowMeter (the tool that generated this CSV) that produces impossible negative numbers and divide-by-zero errors when flow duration is very short or zero.

scanned every numeric column comprehensively (not just guessing which ones), found all affected columns, and are now dropping all affected rows.

```python
# check how many rows have a Flow Duration of exactly 0??

(df['Flow Duration'] == 0).sum()
```

So there can be flow duration zero but the Flow Bytes/s and Flow Packets/s depends on the flow duration.
Dividing by zero gives you inf or NaN , both of which are useless to my model.


### checking the WHOLE dataset properly

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
neg_counts = (df[numeric_cols] < 0).sum()
neg_counts[neg_counts > 0].sort_values(ascending=False)
```

Flow IAT Min, Fwd Header Length, Fwd Header Length.1, min_seg_size_forward, Bwd Header Length, Flow Duration, Flow Packets/s, Flow IAT Mean, Flow IAT Max, Flow Bytes/s  these ARE real problems

### cleaning up

```python
df['Init_Win_bytes_forward'].value_counts().head()
```

Initial Window bytes forward means it's the initial TCP window size.
-1 appears 224,495 times as a single clean repeated value its a placeholder

```python
# columns that should never be negative (excluding Init_Win_bytes, which uses -1 as a valid marker)
bad_columns = [
    'Flow IAT Min',
    'Fwd Header Length',
    'Fwd Header Length.1',
    'min_seg_size_forward',
    'Bwd Header Length',
    'Flow Duration',
    'Flow Packets/s',
    'Flow IAT Mean',
    'Flow IAT Max',
    'Flow Bytes/s'
]

# mark a row as bad if ANY of these columns is negative
row_is_bad = (df[bad_columns] < 0).any(axis=1)

# how many rows are affected
row_is_bad.sum()
```

535 rows are bad in these 10 columns

```python
# condition 1: columns with the negative values
row_is_bad_negative = (df[bad_columns] < 0).any(axis=1)

# condition 2: flow duration is exactly 0 
row_is_bad_zero_duration = (df['Flow Duration'] == 0)

# a row is bad if EITHER condition is true
row_is_bad_final = row_is_bad_negative | row_is_bad_zero_duration

# how many rows total will be dropped
row_is_bad_final.sum()
```

972 rows total are bad this number is less than 535+437=972
actually it equals it exactly here, meaning no row overlapped both problems at once every bad row was bad for only one of the two reasons, not both

## removing the bad rows

```python
df_clean = df[~row_is_bad_final].copy()
df_clean.shape
```

I started with 	529,918 raw rows then removed Negative values + zero duration(972)bug in CICFlowMeter as divide by zero would give me inf

```python
def find_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers, lower_bound, upper_bound
```

```python
outliers_monday, low, high = find_outliers(df_clean, 'Flow Duration')
print(f"Normal range: {low} to {high}")
print(f"Number of outliers: {len(outliers_monday)}")
print(f"Percent of data: {len(outliers_monday)/len(df_clean)*100:.2f}%")
```
